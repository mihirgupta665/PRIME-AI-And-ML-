```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(42)    # now it make the weights and bias initialization reproducible
```

```python
inputs = torch.tensor([1.0, 2.0, 3.0])
inputs
```

```python
neuron = nn.Linear(in_features=3, out_features=1)   # creates a neuron capable of calculating the Weighted Sum
neuron
```

```python
print("Weights : ",neuron.weight)
print(f"Bias : {neuron.bias}")
```

```python
output = neuron(inputs)
output
```

```python
print("verify by input*weight + bias")
print(neuron.weight@inputs + neuron.bias)
```
