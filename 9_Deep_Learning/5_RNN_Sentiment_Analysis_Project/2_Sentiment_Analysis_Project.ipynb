```python
import pandas as pd
import numpy as np

import re
```

```python
df = pd.read_csv("IMDB Dataset.csv")
df
```

```python
print("Shape => ",df.shape)
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

```python
# Lower str
df["review"].str.lower()
```

```python
# Removing url
def remove_urls(text):
    text = re.sub(r"https\S+", "", text)  # (pattern, repl, string) eg: https://www.google.com
    return text
df["review"] = df["review"].apply(remove_urls)
print(df.shape)
df["review"]
```

```python
# removing punctuations
def remove_punctuations(text):
    text = re.sub(r"[^A-Za-z0-9\s]", "", text)   # [] -> excluding this 
    return text

df["review"] = df["review"].apply(remove_punctuations)
df["review"]
```

```python
# removing html
# . -> any one character must be present , * -> means zero or more character exists
def remove_html(text):
    text = re.sub(r"<.*?>", "", text)
    return text
df["review"] = df["review"].apply(remove_html)
df["review"]
```

```python
import nltk

# nltk.download("punkt")
# nltk.download("punkt_tab")
# nltk.download("stopwords")

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    stop_words = stopwords.words("english")
    for word in tokens:
        if word in stop_words:
            text = text.replace(word, "")

    return text

df["review"] = df["review"].apply(remove_stopwords)
```

```python
df
```

```python
def stemming(text):
    stemmed_words = []
    ps = PorterStemmer()
    tokens = word_tokenize(text)
    for token in tokens:
        stemmed_token = ps.stem(token)
        stemmed_words.append(stemmed_token)  # appending each stemmedtoken to list of stemmed word 

    return " ".join(stemmed_words)  # converting list to string using join

df["review"] = df["review"].apply(stemming)
df
```

```python
# batch, sequence and input_size
class RNN(nn.Module):
    def __init__(self, input_size, hidden_size=128, num_layers=1):
        super().__init__()
        self.hidden_size=hidden_size
        self.num_layers=num_layers

        # RNN is present inside the torch.nn
        self.rnn = nn.RNN(input_size, hidden_size, num_layers, batch_first=True)

        self.fc = nn.Linear(hidden_size, 1)

    
    def forward(self, x):

        # h0 hidden size
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size)

        # RNN : 2 outputs, 1> total hiddensize of all time steps and 2> final hiddensize of last timestep only
        out, _  = self.rnn(x, h0)   # dimensions of out => (batch_size, seq_len, hidden_size)

        # print("out : ",out)
        # print("out last : ",out[:,-1, :])
        
        # fc
        out = self.fc(out[:, -1, :])
        return out
```
