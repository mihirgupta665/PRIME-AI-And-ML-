## Merge function:
on : decides which column will be considered for joining the two table;   how : decides the way of joining inner(by default), left, right, outer
- pd.merge("column_name1","column_name2", on="column_name")  # by defult inner join which means only common value in both table forms the new table
- pd.merge("column_name1","column_name", on="column_name", how="left"  ) : how = "way_of_join" , on="column_name":decides whicih column to consider for joining
- inner: only common existing value are the part of joined data
- left/right: all the data points of left/right forms the table and NaN values are given to right/left table unknown values
- outer: all the values of the of both the table forms the new table and un

```python
import pandas as pd
```

```python
df_customers = pd.DataFrame({
    "customer_id" : [1,2,3,4],
    "name" : ["Adam", "Bob", "Charlie", "Dave"]
})
```

## Concatenation of DataFrame
- pd.concat([df1, df2])  : Concatenation takes place by staking one dataframe on top of the other dataframe, labels are preserved.
- pd.concat([df1, df2], ignore_index=True, axis = 1)  : Ignores the label preservement. axis=1 : concatenates the DataFrames side wise(wide or columns wise)
- 

```python
df1 = pd.DataFrame({
    "id": [1,2,3],
    "name": ["Mihir","Vaishali","Divyansh"]
})
df2 = pd.DataFrame({
    "id" : [4,5,6],
    "name" : ["Carles","Darvin","Robert"]
})
```

```python
pd.concat([df1,df2])  # concat two DataFrame by staking on top of the other
```

```python
pd.concat([df1,df2], axis=1)  # axis=1 concatenates column wise
```

```python
pd.concat([df1,df2], ignore_index=True, axis=1)    # ignore_index: it creases the preservement of labelsc
```
