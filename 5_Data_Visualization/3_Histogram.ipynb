# Histogram
It shows a distribution of numerical data, by dividing values into bins(Ranges), and giving the count of values(frequency) which fall into each bin
- seed fixes the starting point of randomness and generates same sequence of random data making it deterministic and reproducible.
## np.random.normal(mean, std_dev, count_of_values) : creates normal distribution of the specified characteristics
plt.hist(data, color="color_name", edgecolor="black") : plots the histogram for the data
- bins: defines the total number of divided ranges (equal and continous), custom bin => bins = list_of_range
## plt.axvline(x-axis, label, color, linewidth, linestyle) : draws a vertical reference line on the plot 


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(0)
scores = np.random.normal(70,10,100)
# print(scores)
bins = [0, 30, 70, 90, 100]
plt.hist(scores, color="green", edgecolor="black", bins=bins)
plt.title("Distribution of marks for students")
plt.xlabel("scores")
plt.ylabel("No. of Student")
plt.axvline(x=33, label="Passing marks (33)", color="red", linewidth= 2, linestyle="--" )
plt.show()
```

```python
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(legit_transactions, bins=10, color="green", edgecolor="black", alpha=0.5, label="Legit Transaction")
plt.hist(fraud_transactions, bins=10, color="orange", edgecolor="black", alpha=0.5, label="Fraud Transaction") 
plt.grid()
plt.xlabel("Transaction amount (in $)")
plt.ylabel("Frequency")
plt.legend()
plt.title("Legit Transaction VS Fraud Transaction")
```

# Box Plots 
## It is a statistical visualization that summarizes the distribution of a dataset using five key numbers:
  1. Minimum(lowest non-outlier) { upper whisker = Q3 + (1.5 * IQR)}  
  2. First Quartile (Q1) - 25 percenttile
  3. Median(Q2) - 50th percentile
  4. Third Quartile - 75 percentile
  5. Maximum (highest non-outlier) { lower whisker = Q1 - (1.5 * IQR)}
- note: IQR(Inter Quartile Range) : { Q3 - Q1 } and 1.5 is the default  whisker length
- It one box plot is made then its x-axis is 1 by default
## plt.boxplot(list_of_data, labels=[list_of_labels], vert=False, showmean, whis=numeric_value)
- vert = True : makes the box plot in vertical direction
- showmean = True : represents the mean in the box plot
- whis = numeric_value : changes the whisker length which is by defaut 1.5
     

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15, 20 ]
plt.boxplot(data, vert=False)
plt.grid(True)
```

```python
group1 = np.random.normal(50,10,100)
group2 = np.random.normal(60,15,100)
plt.boxplot([group1, group2], tick_labels=["group1", "group2"], showmeans=True, whis=1.7)
plt.grid()
plt.show()
```

# Stack Plot
type of plot where multiple data series are stacked on top of each other
each "layer" shows the contribution of one category and all layers are cumulative in nature


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

plt.stackplot(days, direct, organic, social, labels=["direct","organic","social"])
plt.title("Marketing Data")
plt.xlabel("no. of customers")
plt.ylabel("Days")
plt.legend()
```
