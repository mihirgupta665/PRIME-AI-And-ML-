# Histogram
It shows a distribution of numerical data, by dividing values into bins(Ranges), and giving the count of values which fall into each bin
- seed fixes the starting point of randomness and generates same sequence of random data making it deterministic and reproducible.
## np.random.normal(mean, std_dev, count_of_values) : creates normal distribution of the specified characteristics
plt.hist(data,color="") : plots the histogram for the data


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(0)
scores = np.random.normal(70,10,100)
# print(scores)
plt.hist(scores)
plt.show()
```
