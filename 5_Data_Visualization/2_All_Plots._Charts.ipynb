- alignment :  ha="center"
## plt.xlim(lower, upper) : set the limit for lower and upper for x axis
## plt.ylim(lower, upper) : sets the limit for lower and upper for y axis

```python
import matplotlib.pyplot as plt
```

```python

oscar_movies = ["Birdman", "Spotlight", "Moonlight", "The Shape of Water", "Green Book", 
                "Parasite", "Nomadland", "CODA", "Everything Everywhere All at Once", "Oppenheimer"]

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
oscar_revenues = [110, 115, 120, 125, 130, 135, 130, 137.1, 143.5, 146.6, 150]
plt.bar(years, oscar_revenues, color="yellow")
plt.title("Oscar Revenue per year")
plt.xlabel("years")
plt.ylabel("Revenue (in $M)")

for i in range(len(years)):
    plt.text(years[i], oscar_revenues[i]+3, str(oscar_revenues[i]), ha="center")  # print the text on top of the bar

plt.xlim
plt.tight_layout()
plt.ylim(0, max(oscar_revenues)+20)
plt.xlim(min(years)-2, max(years)+2)
plt.show()
```
