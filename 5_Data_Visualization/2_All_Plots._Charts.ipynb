- alignment :  ha="center"
## plt.xlim(lower, upper) : changes the limit for lower and upper for x axis
## plt.ylim(lower, upper) : changes the limit for lower and upper for y axis

```python
import matplotlib.pyplot as plt
```

```python

oscar_movies = ["Birdman", "Spotlight", "Moonlight", "The Shape of Water", "Green Book", 
                "Parasite", "Nomadland", "CODA", "Everything Everywhere All at Once", "Oppenheimer"]

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
oscar_revenues = [110, 115, 120, 125, 130, 135, 130, 137.1, 143.5, 146.6, 150]
plt.bar(years, oscar_revenues, color="yellow")
plt.title("Oscar Revenue per year")
plt.xlabel("years")
plt.ylabel("Revenue (in $M)")

for i in range(len(years)):
    plt.text(years[i], oscar_revenues[i]+3, str(oscar_revenues[i]), ha="center")  # print the text on top of the bar

plt.tight_layout()
plt.ylim(0, max(oscar_revenues)+20)
plt.xlim(min(years)-2, max(years)+2)
plt.show()
```

- default width of each graph is 0.8
- width decrease to 0.4 to plot 2 and 0.27 to plot 3
- and x-width/2, x+width/2 or x-width,x,x+width written in plot to have width

```python
import numpy as np

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

oscar_movies = ["Birdman", "Spotlight", "Moonlight", "The Shape of Water", "Green Book", "Parasite",
                "Nomadland", "CODA", "Everything Everywhere All at Once", "Oppenheimer"]
oscar_revenues = [103, 98, 65, 195, 321, 262, 39, 2, 141, 975]

non_oscar_movies = ["Mad Max: Fury Road", "The Revenant", "La La Land", "Dunkirk", "Bohemian Rhapsody",
                    "Joker", "Tenet", "Dune", "Top Gun: Maverick", "Avatar: The Way of Water"]
non_oscar_revenues = [375, 533, 448, 527, 910, 1074, 365, 402, 1496, 2320]


x = np.arange(len(years))
width = 0.4


plt.bar(x-width/2, oscar_revenues, width, label="Oscar Revenue")
plt.bar(x+width/2, non_oscar_revenues, width, label="Non-Oscar Revenue")
plt.legend()
plt.title("Oscar Revenue Vs Non-Oscar Revenue")
plt.tight_layout()
# plt.xlim(0-1, 10+1)
plt.show()
```

```python
oscar_movies = ["Birdman", "Spotlight", "Moonlight", "The Shape of Water", "Green Book", "Parasite",
                "Nomadland", "CODA", "Everything Everywhere All at Once", "Oppenheimer"]
oscar_revenues = [103, 98, 65, 195, 321, 262, 39, 2, 141, 975]
plt.barh(oscar_movies, oscar_revenues)  # y-axis, x-axis unlike bar(X,Y)  barh is barh(Y,X)
plt.title("Movies  and thier revenue")
plt.xlabel("Revenue")
plt.ylabel("Movie Names")
```
