# Matplotlib : lib for creating static, animated and interactive visualization
## matplotlib.pyplot : pyplot is the toolkit for drawing
## plt.plot(X,Y) : plot a line chart and x,y need to be passed as parameter
- plt.plot(X,Y, label="name") : is to to define the plot(represents what) label
## plt.show() : it is used to display any kind of chart or graph on the screen 
## plt.legend(loc="location") : displays a legend box with all the plot labels(plot representation), location defines the location of the display of box
- upper left/right or center or lower left/right, in default it is located at the best location by itself

```python
# pip install matplotlib
import matplotlib.pyplot as plt
```

```python
X = [1,2,3,4]
Y = [5,6,7,8]
plt.plot(X,Y)
plt.show()
```

```python
oscar_movies = ["Birdman", "Spotlight", "Moonlight", "The Shape of Water", "Green Book", 
                "Parasite", "Nomadland", "CODA", "Everything Everywhere All at Once", "Oppenheimer"]

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
oscar_revenues = [110, 115, 120, 125, 130, 135, 130, 137.1, 143.5, 146.6, 150]
plt.plot(years, oscar_revenues)
plt.title("Oscar Movie Revenue per year")
plt.xlabel("years")
plt.ylabel("Revenue (in $M)")
```

```python
non_oscar_movies = ["Mad Max: Fury Road", "The Revenant", "La La Land", "Dunkirk", 
                    "Bohemian Rhapsody", "Joker", "Tenet", "Dune", "Top Gun: Maverick", "Oppenheimer", "Dune: Part Two"]

non_oscar_years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
non_oscar_revenues = [375, 533, 448, 527, 910, 1074, 365, 402, 1496, 975, 820]

plt.plot(years, oscar_revenues, "^--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenues, ">:y", label="Non-Oscar Movies")
plt.title("oscar movie revenue VS non-oscar movie revenue")
plt.xlabel("years")
plt.ylabel("Revenue   (in $M)")
plt.legend(loc="upper left")
```

## with plt.xkcd(): plots written in this block are cartoonish plots
## Format String : it is a string which consists of a part for color, marker, line
fmt = '[market][line][color]'  eg: o--r
## plt.plot(X,Y, color="name", marker="symbol", linestyle="style", linewidth="numerical_value")
## plt.style.available  : shows all the styles which is available and could be applied to the specific plot
- plt.style.use("attribute_name") : from the list we could choose appropriate style and use them    ("default"  : to bring back normal style for graph)
## plt.grid()  : is used to display grid in the background
## plt.tight_layout() : is used to have good spacing for the plot, prevents overlapping too
## plt.savefig("name_plot.png") : save the plot in the pnng format with the specific name
  

```python
plt.plot(years, oscar_revenues, "^--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenues, color="purple", marker="o", linestyle="-.", linewidth=5, label="Non-Oscar Movies")
plt.title("oscar movie revenue VS non-oscar movie revenue")
plt.xlabel("years")
plt.ylabel("Revenue   (in $M)")
plt.legend(loc="upper left")
print(plt.style.available)
plt.style.use("dark_background")
plt.grid()
plt.tight_layout()
```

```python
with plt.xkcd():
    plt.plot(years, oscar_revenues, label="Oscar Movies")
    plt.plot(years, non_oscar_revenues, label="Non-Oscar Movies")
    plt.style.use("default")
    plt.title("oscar movie revenue VS non-oscar movie revenue")
    plt.xlabel("years")
    plt.ylabel("Revenue   (in $M)")
    plt.legend(loc="upper left")
    
```
