## plt. -> was a state based AP
# object based api is the modern matplotlib
# fig, ax = subplots(nrow, ncol, figsize=(size,size)) : returns two object fig, ax
- empty subplots() : creates only one plot (not multiple subplots)
- figsize=(numerical_value, numerical_value) : sets the size for individual graph(ax)
- fig is the entire canvas where all the plots are drawn together
- ax is the 2-D array of individual plot
- ax[i][j].plot(X,Y)  : to plot the individual graph
- ax[i][j].set_title("title_name") : to set the title for the individual graph
- ax[i][j].set_xlabel("label")/.set_ylabel("label") : sets the xlabel and ylabel for the indiviual plots
- fig.tight_layout() : sets a clear layout of the figure
- fig.suptitle("Name_of_Figure") : sets suptitle for the entire figure
- fig.supxlabel("label")/fig.supylabel("label") : generates the x and y label for the intire plot 
- 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(10)+1
y1 = np.sqrt(x)
y2 = x*2
y3 = x**2
y4 = x**3
y4
```

```python
fig, ax = plt.subplots(2, 2, figsize=(7,7))
ax[0][0].plot(x, y1)
ax[0][0].set_title("plot1 - Sqrt")
ax[0][0].set_xlabel("X-Axis")
ax[0][0].set_ylabel("Y-Axis")
ax[0][1].plot(x, y2)
ax[0][1].set_title("plot2 - Double")
ax[1][0].plot(x, y3)
ax[1][0].set_title("plot3 - Square")
ax[1][1].plot(x,y4)
ax[1][1].set_title("plot4 - Cube")

fig.suptitle("Sub Plots")
fig.tight_layout()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [25, 26, 24, 27, 28]
]

fig, ax = plt.subplots(2,2, figsize=(7,7))
city_no = 0
for i in range(2):
    for j in range(2):
        ax[i][j].plot(days, temperatures[city_no], marker="o")
        ax[i][j].set_title(cities[city_no])
        ax[i][j].grid(True)
        ax[i][j].set_xlabel("Days")
        ax[i][j].set_ylabel("Temperature")
        city_no = city_no+1

fig.suptitle("Differnt City with thier temperature")
fig.supxlabel("Days")
fig.supylabel("Temperature")
fig.tight_layout()
```

## (2Darray).flatten() : converts the array into 1-D array
## for i,val in enumerate(axes): enumerates here gives the index and the corresponding value 

```python
fig, axes = plt.subplots(2,2)
axes = axes.flatten()
for i,ax in enumerate(axes):
    ax.plot(days, temperatures[i], marker="o")
    ax.set_title(cities[i])
    ax.grid(True)

fig.suptitle("Cities with their temperature")
fig.supxlabel("Days")
fig.supylabel("Temperature")
fig.tight_layout()
```
