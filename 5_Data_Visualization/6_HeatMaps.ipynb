 # HeatMaps
 ## Plots rectangular data as color-encoded matrix
sns.heatmap(
    data,
    cmap="style_name",
    annot=True/False,
    fmt="Format_Datatype"
)
- data.pivot(index=row, columns=[list_of_columns] values="column_name" )
  index creates row, columns creates list of columns and values are the values of the cells formed
- cmap : gives the style of the heat map
- annot : if True writes about the information of each cell 
- fmt : formats the data type of the information mentioned by the annot, "d" for integer

```python
import  matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights = sns.load_dataset("flights")
flights
```

```python
flights_pivot = flights.pivot(index="month",columns="year",values="passengers")
flights_pivot
```

```python
sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
plt.title("Passengers HeatMap")
plt.tight_layout()
```

# Best Practices
- Create the moder object subplots of matplotlib and then create the individual plots by seaborn
- errorbar=None : to hide the confidence interval
- ax is a optional option for selection of axes to draw the plot

```python
fig, ax = plt.subplots()
tips = sns.load_dataset("tips")
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax,
)
ax.set_title("Avg total Bill amount for each Day")
fig.tight_layout()
```
