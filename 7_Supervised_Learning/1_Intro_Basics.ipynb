# Machine Learning
- ML is a part of computer science where we teach computers to learn from data, find pattern and make decision or predictions.

# Supervised Learning :
- Supervised ML trains algorithms on labelled datasets, meaning each input has a known correct output, to learn patterns(or hypothesis or relationship) and predict outcomes for new data 
### Labelled Data
- It is a raw data (img, txt or audio) which are paired with meaningful labels(or tags or annotation) that informs the machine learning model about the desired outcome.

## sklearn
- it is a open source ml library that supports supervised and unsupervised learning



# Linear Regression
h(x) = A + B.X1 + C.X2 + D.X3 + E.X4
- A is bias or intercept and B is parameter or coefficient
- best fit line is the line whose predictions are very near to actual outputs thereby reducing the errors.



## Cost Function
- Measures how well the model's predictions match the actual data.  [ h(X_i) - y_i ]
## Gradient Descent
- Gradient Descent is an iterative optimization algorithm used to minimize a cost function by adjusting model parameters in the direction of the steepest descent of the function's gradient. [T_k = T_k - a.(dy/dt) ]
