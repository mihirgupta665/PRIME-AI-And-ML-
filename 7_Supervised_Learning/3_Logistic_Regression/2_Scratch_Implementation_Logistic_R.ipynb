```python
import numpy as np
```

```python
class Logitic_Regression:
    def __init__(self, learning_rate=0.1, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def _sigmoid(z):
        return (1 / (1 + np.exp(-z)))

    def fit(self, X, y):
        m, n = X.shape
        
        self.bias = 0
        self.weights = np.zeroes(n)

        for i in range(self.n_iter):
            
            z = self.bias + np.dot(X.T, self.weights)
            y_pred = self._sigmoid(z)

            db = (1/m) * np.sum(y_pred - y)
            dw = (1/m) * np.dot(X.T, np.sum(y_pred - y))

            self.bias -= self.lr*db
            self.weights -= self.lr*dw
    
    def get_probabilities(self, X):
        z = self.bias + np.dot(X.T, self.weights)
        probability = self._sigmoid(z)
        return probability

    def predict(self, X, threshold=0.5):
        probability = self.get_probabilities(X)
        y_pred_bool = probability >= threshold
        return y_pred_bool.astype(int)
        

    
        
```
