```python
import pandas as pd
from sklearn import datasets
from sklearn.metrics import r2_score
from sklearn.svm import SVR
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
```

```python
df = datasets.load_diabetes(as_frame=True).frame
df
```

```python
X = df.drop("target", axis=1)
y = df["target"]
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
X_train.head()
```

```python
y_scaler = StandardScaler()
y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()  # y is 1-D and fiting is done on 2-D
y_test_scaled = y_scaler.transform(y_test.values.reshape(-1,1)).ravel()
y_train.head()
```

```python
model = SVR()
model.fit(X_train, y_train_scaled)
y_train_pred_scaled = model.predict(X_train)
y_test_pred_scaled = model.predict(X_test)
print("Train R2 Score : ",r2_score(y_train_scaled, y_train_pred_scaled))
print("Test R2 Score : ",r2_score(y_test_scaled, y_test_pred_scaled))
```

```python
model = SVR(kernel="linear")
model.fit(X_train, y_train_scaled)
y_train_pred_scaled = model.predict(X_train)
y_test_pred_scaled = model.predict(X_test)
print("Train R2 Score : ",r2_score(y_train_scaled, y_train_pred_scaled))
print("Test R2 Score : ",r2_score(y_test_scaled, y_test_pred_scaled))
```

```python
model = SVR(kernel="poly")
model.fit(X_train, y_train_scaled)
y_train_pred_scaled = model.predict(X_train)
y_test_pred_scaled = model.predict(X_test)
print("Train R2 Score : ",r2_score(y_train_scaled, y_train_pred_scaled))
print("Test R2 Score : ",r2_score(y_test_scaled, y_test_pred_scaled))
```

```python
# Grid Search CV
param_grid = {
    "C": [1, 1.5, 2, 3, 5, 10, 15, 20, 25],
    "kernel": ["rbf", "linear", "poly"],
    "epsilon": [0.01, 0.05, 0.1, 0.2, 0.3, 0.5, 0.7]

}
```

```python
svr = SVR()

grid_search = GridSearchCV(
    svr,
    param_grid,
    scoring="r2",
    cv=5
)

grid_search.fit(X_train, y_train_scaled)
```

```python
print("Best Params : ",grid_search.best_params_)
```

```python
best_model = SVR(C=15, epsilon=0.1, kernel="linear")

best_model.fit(X_train, y_train_scaled)
y_train_pred_scaled = best_model.predict(X_train)
y_test_pred_scaled = best_model.predict(X_test)
print("Train R2 Score : ",r2_score(y_train_scaled, y_train_pred_scaled))
print("Test R2 Score : ",r2_score(y_test_scaled, y_test_pred_scaled))
```
