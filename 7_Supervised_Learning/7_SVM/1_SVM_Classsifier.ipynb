- C.sigma_1_w(E.e_i) : slack penalty,  E - Slack variable i.e degree of violation calculated as the sum of distanceof misclassified point from the margin hyperplane

```python
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
```

```python
df = datasets.load_iris(as_frame=True).frame
df
```

```python
df.isnull().sum()
```

```python
X = df.drop("target", axis=1)
y = df["target"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
X_train.head()
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
svm =  SVC()
svm.fit(X_train_scaled, y_train)
y_pred = svm.predict(X_test_scaled)
print(f'Accuracy Score : {accuracy_score(y_test, y_pred)}')
# print(f'F1 Score : {f1_score(y_test, y_pred)}')
print("Classification Report\n", classification_report(y_test, y_pred))
```

```python
# Linear Kernel
svm = SVC(kernel="linear")
svm.fit(X_train_scaled, y_train)
y_pred = svm.predict(X_test_scaled)
print(f"Accuracy Score : {accuracy_score(y_test, y_pred)}")
```

```python
# Linear Kernel
svm = SVC(kernel="poly")
svm.fit(X_train_scaled, y_train)
y_pred = svm.predict(X_test_scaled)
print(f"Accuracy Score : {accuracy_score(y_test, y_pred)}")
```

```python
# Sigmoid Kernel
svm = SVC(kernel="sigmoid")
svm.fit(X_train_scaled, y_train)
y_pred = svm.predict(X_test_scaled)
print(f"Accuracy Score : {accuracy_score(y_test, y_pred)}")
```

```python
# linear HyperParameter tuning
c_vals = [0.125, 0.25, 0.3, 0.5, 0.7, 1, 1.2, 1.5, 2, 3, 4, 5]
for c_val in c_vals:
    svm = SVC(C=c_val, kernel="rbf")
    svm.fit(X_train_scaled, y_train)
    y_pred = svm.predict(X_test_scaled)
    print(f"AT C = {c_val} we get Accuracy Score : {accuracy_score(y_test, y_pred)}")
    
```
