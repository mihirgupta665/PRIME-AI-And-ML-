# Scratch implementation of Linear Regression
## Initialization:
- Bias
- Parametric Weights
- Learning Rate
- Number of iterations

```python
import numpy as np
```

```python
class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter
        
    def fit(self, X, y):
        m, n = X.shape   # return rows(samples) and columns(features) of a array

        # Step-1 : Proper Initialization
        self.bias = 0,
        self.weights = np.zeros(n)   # return a array with each value as 0

        for i in range(self.n_iter):
            # Step-2 : Y_Prediction
            y_pred = self.bias + np.dot(X, self.weights)
    
            # Step-3 : derivatives of bias and weights
            db = (1/m) * np.sum(y_pred - y)
            dw = (1/m) * np.dot(X.T, (y_pred - y))
    
            #Step-4 : bias and weights updation apply convergence theorem 
            self.bias -=  self.lr * db
            self.weights -= self.lr * dw
        
   
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2,4,6,8,10])

model = LinearRegression()
model.fit(X,y)
y_pred = model.predict(X)
print(y_pred)
print(model.bias)
print(model.weights)
```
