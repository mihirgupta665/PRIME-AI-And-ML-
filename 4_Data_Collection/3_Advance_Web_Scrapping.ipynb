```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)
    soup = BeautifulSoup(res.text, 'lxml')
    quotes = soup.select("div.quote")   # diffence need to get the obeservation whether page exists or not
    # print(quotes)
    if not quotes:
        print("No qoutes at this page",page_count)
        break

    with open(f"scrapped_data/quotes{page_count}.html", "w", encoding="utf-8") as f:    # to write html text encoding="utf-8" need to be passed as p
        # print(res)
        f.write(res.text)
        print("Downloaded data from page",page_count)
        
    page_count+=1
```
