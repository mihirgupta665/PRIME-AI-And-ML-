```python
from bs4 import BeautifulSoup

with open ("scrapped_data/data1.html", "r") as f:
    html_content = f.read()
```

```python
soup = BeautifulSoup(html_content, "lxml")
# soup
```

# Soup Methods
## Soup.find('element_name') or soup.element_name : they both return the first occurance of element
## Soup.find_all('element_name') : returns all the occurance of that element
## element.get_text() : is used to get the content of the element

```python
soup.find('h3')  # returns the first occurence of the element
```
