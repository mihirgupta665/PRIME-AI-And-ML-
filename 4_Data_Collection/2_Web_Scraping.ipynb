```python
from bs4 import BeautifulSoup

with open ("scrapped_data/data1.html", "r") as f:
    html_content = f.read()
```

```python
soup = BeautifulSoup(html_content, "lxml")
# soup
```

# Soup Methods
## Soup.find('element_name') or soup.element_name : they both return the first occurance of element
## Soup.find_all('element_name') : returns all the occurance of that element
## element.get_text(strip=True) : is used to get the content of the element, strip parameter removes the spaces
## element.find_parent("tag") : is used to get the specified tag parent of the element
- element.find_parent("tag")["attributs"] : is used to get attribute(class) value the specified tag parent of the element
## element.childern() : is used to get the LIST of children of the element
## element.find_next() : finds the next sibling
## element.find_previous() : finds the previous sibling
## element.select("tag.class") : it directly finds and return the desired element in form of list
## element.select_one("tag.class") : it returns the first instance of the selected list
## element.get() or element.attrs : returns all the attributes of the same class
## element.descendants : returns the list of all the descendants of the element
## next_sibling(), previous_sibling() : to navigate to the next or previous sibling 

```python
soup.find('h3')  # returns the first occurence of the element
```

```python
soup.h3
```

```python
all_h3 = soup.find_all('h3')
# all_h3
type(all_h3)
```

```python
countries_population = []
for h3 in all_h3:
    # print(h3.get_text(strip=True))
    name = h3.get_text(strip=True)
    # print(h3.find_parent("div")["class"])
    # print(h3.find_next("div").select("span.country-population")[0].get_text(strip=True))
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)

    countries_population.append([name, population])
```

```python
countries_population
```

```python
import pandas as pd
df = pd.DataFrame(countries_population, columns=["Name", "Population"])
df
```

```python
df.to_csv("cleaned_data/data1.csv", index=False)  # df.to_csv() : is used to save dtaframe to csv and index parameter decides to save index or not
```
