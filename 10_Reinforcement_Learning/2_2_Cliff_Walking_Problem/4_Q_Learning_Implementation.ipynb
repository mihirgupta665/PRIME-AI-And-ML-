# Q Learning : Off policy (finds the optimal action to take at each next step)

```python
import gymnasium as gym
import numpy as np
import random
```

```python
alpha = 0.5
gamma = 0.99
epsilon = 0.1
episodes = 500
```

```python
Q = np.zeros((48,4))
```

```python
# Policy for intial step
def epsilon_greedy(state):
    if(random.random() < epsilon):
        return env.action_space.sample()  # exploration
    else:
        return np.argmax(Q[state])
```

```python
# training the Agent

for episode in range(episodes):
    
    if (episode % 50 == 0 and episode !=0 and episode!=499):
        env = gym.make("CliffWalking-v1", render_mode="human")
    else:
        env = gym.make("CliffWalking-v1")

    total_reward = 0
    episode_len = 0
    done = False

    state, _ = env.reset()

    while not done:
        action = epsilon_greedy(state)
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        # Q Learning update
        Q[state, action] = Q[state, action] + (alpha * (reward + gamma * np.max(Q[next_state]) - Q[state, action]))
        state = next_state
        
        total_reward += reward
        episode_len += 1


    print(f"Episode = {episode} with episode length = {episode_len} and total reward = {total_reward}")
    env.close()
        
    
```
