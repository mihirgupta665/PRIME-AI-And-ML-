# SARSA

```python
gamma = 0.99
alpha = 0.5
epsilon = 0.1
episodes = 650
```

```python
# Q table
Q = np.zeros((48,4))   # total state = 48
```

```python
# Policy
def epsilon_greedy(state):
    if (random.random() < epsilon):
        return env.action_space.sample()  # explore
    else:
        return np.argmax(Q[state])  # exploit
```

```python
for episode in range(episodes):
    
    render = (episode % 50 == 0 and episode != 0 and episode != 649)
    if render : 
        env = gym.make("CliffWalking-v1", render_mode="human")     # env mode creation
        print(f"Human Level Rendering for the Episode = {episode+1}")
    else:
        env = gym.make("CliffWalking-v1")

    done = False      # check if terminated or truncated
    total_reward = 0
    episode_len = 0

    state, _ = env.reset()   # gives 2 values state and some information
    # print(env.reser())
    action = epsilon_greedy(state)  # on policy SARSA decides the action on basis of policy

    while not done:    # till done = True

        next_state, reward, terminated, truncated, _ = env.step(action)  # gives 5 parameters

        done = truncated or terminated
        next_action = epsilon_greedy(next_state)  # on policy so action is decided by policy

        # SARSA updation of Q values of Q tables
        Q[state, action] = Q[state, action] + (alpha * (reward + gamma * (Q[next_state, next_action] - Q[state, action])))

        total_reward += reward   # reward cumulating
        episode_len += 1

        # reached to new state so state changed and action changed
        state = next_state   
        action = next_action


    print(f"episode : {episode+1} has total length = {episode_len} and reward = {total_reward}")
    env.close()  # close that episodes environment
```

```python
# Evaluation : What are agent learned 

env = gym.make("CliffWalking-v1", render_mode="human")

state, _ = env.reset()
done = False
total_reward = 0
episode_len = 0

while not done:
    action = np.argmax(Q[state])
    next_state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated

    total_reward += reward
    episode_len += 1

    state = next_state

print(f"Total Reward = {total_reward} and Episode Length = {episode_len}")
env.close()
```
