## Properties of Numpy
- Numpy are are faster as they are optimised in C
- Numpy store data in continuous memory block
- Uses less memory (optimised storage)
- Built in Mathematical Functions
- Supports direct vectorised operations (operation execute in bulk)

# Time Performace comparision


```python
import time
size = 100_000_000
py_list = list(range(size))  # creates a list of specified range

start1 = time.time()
sq_list = [val**2 for val in py_list]    # list comprehension has expression in beginning
end1 = time.time()
print(f"Total time for Squaring Python List is : {end1-start1}")

np_arr = np.array(py_list)  # its takse a parameter list and converts it into array capable of vectorization
start2 = time.time()
sq_arr = np_arr ** 2 
end2 = time.time()
print(f"Total time for Squaring Numpy Array is : {end2-start2}")
```

# Space Peformance Comparision

```python
import sys

list_size = sys.getsizeof(py_list) 
elements_size = sum(sys.getsizeof(x) for x in py_list)
total_size = list_size + elements_size
print(f"Size of py_list is : {total_size}")

print(f"Size of nb_array is : {np_arr.nbytes}")
```
