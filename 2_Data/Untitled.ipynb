```python
import json
```

```python
# loading the data
def load_data(filename):
    with open(filename, "r") as f : 
        data = json.load(f)    # file object could only be operated
    return data
data = load_data("store_data.json")
print(type(data))
print(data)
    
```

# Clean and Structure the data
## Clean and Consistent
## Missing Values Fullfillment
## Deduplication

```python
def clean_data(data):
    unique_users = set()
    cleaned_data = []
    str_to_num = {"one" : 1, "two" : 2, "three" : 3, "four" : 4, "five" : 5}

    for user in data:
        # clean and consitent rating
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in str_to_num):
            raw_rating = str_to_num[raw_rating]  # raw_rating contains the rating for mapping too.
        user["rating"] = raw_rating   # without mapping we want striped data and lower case
        
        # Missing Values Fullfillment
        if(user.get("age")==None):
            user["age"]=None
    
        # Deduplication
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"].strip())
        cleaned_data.append(user)
    
    return cleaned_data
    
    
```

```python
data = clean_data(data)
```

## Valuable Insights

```python
def get_insights(data):
    sum_rating = 0
    poor_rating = 0
    for user in data:
        sum_rating+=float(user["rating"])
        if(float(user["rating"]) < 3):
            poor_rating+=1

    avg_rating = sum_rating/len(data)
    print(f"Average Rating of Product is = {avg_rating}")
    print(f"Percentage of Poor Rating(<3) is = {poor_rating/len(data)*100}% ")
        
```

```python
get_insights(data)
```
