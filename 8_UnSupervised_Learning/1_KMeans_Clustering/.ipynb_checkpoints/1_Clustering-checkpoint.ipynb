# Clustering : 
- It is an unsupervised ML technique that automatically group similar, unlabeled data points into clusters based on defined similarity metrics.
### It is 3 types
- Partitioning Labels -> KMeans Clustering
- Hierarchial Clustering
- Density Based -> DBSCAN

### Elbow method : Sigma_1_K(Sigma_1_n(distance of p_i from centroid)^2)     [S_i_k(Sigma_1_n(p_i - u_k)^2)]
