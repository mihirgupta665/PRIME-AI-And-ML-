```python
from sklearn.datasets import make_blobs
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
X, y = make_blobs(
    n_samples=1000,
    n_features=2,
    centers=4,
    random_state=42
)
X
```

```python
k=4
kmeans = KMeans(
    n_clusters=k,
    random_state=42
)
```

```python
labels = kmeans.fit_predict(X)
labels
```

```python
sns.scatterplot(x=X[:,0], y=X[:,1], c=labels)
```

```python
# Elbow method
```

```python
wcss = []
for k in range(1,21):
    kmeans_elbow = KMeans(
        n_clusters=k,
        random_state=42
    )
    kmeans_elbow.fit_predict(X)
    wcss.append(kmeans_elbow.inertia_)
```

```python
sns.lineplot(x=range(1,21), y=wcss, marker="o")
```
