```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
iris = load_iris()
iris
```

```python
# X= pd.DataFrame(iris.data)
X=iris.data
y= iris.target
```

```python
# X.columns = iris.feature_names
print(X.shape)
X
```

```python
sns.scatterplot(x=X[:,0], y=X[:,2], c=y)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca_data = pca.fit_transform(X_scaled)
X_pca_data
```

```python
# Elbow K
wcss = []
for k in range(1,11):
    kmeans = KMeans(
        n_clusters=k,
        # random_state=42
    )
    kmeans.fit(X_pca_data)
    wcss.append(kmeans.inertia_)
    
```

```python
sns.lineplot(x=range(1,11), y=wcss, marker="o")
```

```python
kmeans = KMeans(
    n_clusters=3,
)
labels = kmeans.fit_predict(X_pca_data)
```

```python
sns.scatterplot(x=X_pca_data[:,0], y=X_pca_data[:,1], c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:,0], y=kmeans.cluster_centers_[:,1], marker="X", c="red", s=150)
```
