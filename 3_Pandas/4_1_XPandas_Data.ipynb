# Pandas' Datastructure : Series and Data Frame
- Series and DataFrame are classes
- Series : is a 1-D Labelled Array
- DataFrame : is a 2-D Labelled Array
- Note : DataFrame is collection of many different vertical Series
- These Pandas' DataStructure understand the data and could arrange headers and row data by themselves with proper labeling at each row.
- index : position    but    label : named index
- labeled are defined in index parameter index = [values]  (labeled is assigned thourgth the parameter label)
- s.index : returns the list of all the labels (not index),    [Note : Label are named Index]

```python
import pandas as pd
```

## Series and Its Properties
- 1-D labelled Array (vertically visualized)
- for any operation between different Series the label must be same else not desired operations is not possible
- Homogenous and Vectorization
- Handle missing values with NaN (same as python : None)
- mutable values but immutable size   (if size is altered than a new copy of series is created)
- .drop(index/label)  : creates a copy of series  with the removed value of the specified index 

```python
# Series
s = pd.Series([11,22,55,44,33], index=["AA", "BB", "CC", "DD", "EE"])   # label is assigned as a list in index parameter 
print(s)
print(s.dtype)
print(f"The value at position 1 is : {s[1]}")
print(f"The label 'CC' has beeen assigned to the value : {s['CC']}")    # through labels value could be extracted
print(s.index)    # returns the list of all the labels(named index)

ss = pd.Series([77,88,99,110,121], index=["AA", "BB", "CC", "DD", "EE"])   # for any operation between Series the label must be same else not possible
print(s + ss)

sx = s.drop("AA")      # size is immutable so copy of series is created with droped value
print(s)
print(sx)

```

## DataFrame And its Properties
- Its datatype is 'pandas.core.frame.DataFrame'
- pd.DataFrame(collection, columns=[values])   : columns parameter assign labels to each column
- df.index : returns the row labels in form of RangeIndex with start, stop, step parameters
- df.columns : returns the column labels

```python
# dictionary with list inside
info = {
    "Name" : ["Mihir", "Vaishali", "Divyansh"],
    "Age" : [23,22,23],
    "CGPA" : [9.8, 8.98, 8.6]
}
df = pd.DataFrame(info)
print(df)
print(type(df))
print(df.index)
print(df.columns)
```

```python
df
```

```python
# List with List and another parameter columns for the column labels(headers)
df1 = pd.DataFrame([["Mihir",  21], ["Vaishali", 20], ["Divyansh", 22]], columns=["Name", "Age"])
print(df1)
print()
# numpy array
import numpy as np
np_arr = np.array([[1,2,3], [4,5,6]])
df2 = pd.DataFrame(np_arr, columns=["A","B","C"])    # labels are passed as list in columns parameter
print(df2)
```
