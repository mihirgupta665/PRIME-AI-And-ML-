# Group by And Aggregation
- grouby(coumn_name) : forms rows as group of values of the specified column but we need a aggregation_funtion() to have some implementation on the groups
- aggregation operation could be : [column_name].min(), .max(), .mean(), .sum() ,.count()
- df.groupby('column_name')[column_name].aggregation_function()
- [column_name].agg([...aggregate_operations...]) or aggregate() : this function is used for implemenation of more than one aggregate operation on the specified column
- df.agg(column_name1=aggregate_operation1, column_name2=aggregate_operation2):assigns specified names to the aggregation operations on entire table
- df.agg(specified_name1=(column_name1,aggregate_operation), specified_name2=(column_name1, aggregation_operation))

```python
import pandas as pd
```

```python
df = pd.read_csv("4_5_Transformed_Data.csv")
```

```python
df2 = df.copy()
df2
```

```python
df2.groupby('Country')['Salary'].mean()   # group by specifies which columns value need to be considered for grouping.
```

```python
df2.groupby('Gender')['Salary'].min()   # [column_name] decides which column need to be considered for implementation of aggregation function.
```

```python

#['column_name'] aggregate() or agg(): to apply many aggregate operations on same column of a data of gnerated by a groupby function.
df.groupby('Country')['Salary'].agg(["mean", "min", "max", "count", "sum"])
```

```python
# renaming of the column could be done by writing {   name="aggregate_operation"    }
df.groupby('Country')['Salary'].agg(Avg_Salary="mean", Min_Salary="min", Max_Salary="max", Country_Count="count", Sum_of_Salary="sum")
```

```python
df2.groupby('Country').agg({  # .agg({})
    'Salary':"mean",
    'Age':"min"
})
```

```python
df2.groupby('Country').agg(
    Avg_Salary=('Salary',"max"),
    Min_Age=('Age',"min")
)
```
