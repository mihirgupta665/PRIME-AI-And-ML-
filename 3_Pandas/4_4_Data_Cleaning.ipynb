# Data Cleaning Using Pandas



## Handling Missing Values :
- all the changes executed on the copy so changes does not affect the original data 
- df.isnull() or isna()   : prints the table will 'True' at those position whose values are missing
- df.isnull().sum() or isna().sum() : prints column wise the count of missing values in each column.
- df.dropna() or dp.dropna(axis=1) : removes the entire row or column respectively where a null value exists
- df.fillna(value)  : assign the specified value to each empty cell
- df.ffill() : from top to bottom of a column empty cell is filled by the last encountered not-null value
- df.bfill() : from bottom to top of a column empty cell is filled by the last encountered not-null value (if first element is missing then NaN is assigned or preserved)

```python
import pandas as pd
```

```python
df = pd.read_csv("4_4_persons_raw_data.csv")
df
```

```python
df.isnull()   # prints True at those positions where values are missing
```

```python
df.isna().sum()   # prints column wise the count of missing values in each column
```

```python
df.dropna()    # prints the row whose every cell is not null
```

```python
df.dropna(axis=1)   # prints the only columns whose every cell is not null
```

```python
df.fillna(0)   # fills each empty cell with specified value
```

```python
cleaned_data =  df.copy()
age_mean = cleaned_data["age"].mean()   # chaining mean function to filtered age column to calculate mean
cleaned_data = cleaned_data["age"].fillna(age_mean)   # assigning the age column's empty value the mean of age to get consistent data of age column 
cleaned_data
```

```python
df.ffill()   # from top to bottom of column the empty cells are filled with the last encountered not-null value
```

```python
df.bfill()  # from bottom to top in a column the empty cells are filled with the last encountered not-null value
```

## Handling the duplicate values : 
- df.duplicated() : returns True for the row with duplicate value (second time True is returned , first time False is returned
if checked on entire table -> if each cell of more than one row are same then those rows are duplicate rows  
- df[[[set_of_column]].duplicated() : checks and return True if all the cells in specified columns for a row is exactly same.

```python
df.duplicated()   # all cells of more than one row must be same for True to be returned
```

```python
df[["country", "gender"]].duplicated()  # selectedcolumns cells need to be same for more than one row for True to be returned
```
