# Melt - wide to Long format
- melt(id_vars=[column_names], value_vars, var_name, value_name)
- id_vars=[...column_names] : list column which is preserved in the table
- value_vars=[...column_names] : list of columns which need ro be merged in one column
- var_name = "String_Name" : name written for the new column containing th merged column info
- value_name = "string_name" : column name for storign the value of which merged column

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110]  ,
    "profit": [20,25,18,22],
    "tax": [10,20,30,40]
})
```

```python
df
```

```python
melted_data = df.melt(
    id_vars=["country", "year"],   # column which is preserved in the table
    value_vars=["sales", "profit", "tax"],  # column which needs to be merged in the one column
    var_name="metrics",  # column name in which old merge column are written
    value_name="value"   #  column value for the corresponding old merged column value
)
melted_data
```

# Pivot - Long to Wide format
- df.pivot(index=[...column_names], column="metrics", values="value")
- index = [...column_names] column or set of column to keep they must be unqiue together for a row
- column = "column_name": name of column to slipt it in different column so that data could be converted to long each value of this column becomes new column
- values="value" : specifies column name in which  of the column to be split values are stored

```python
melted_data.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)
```
