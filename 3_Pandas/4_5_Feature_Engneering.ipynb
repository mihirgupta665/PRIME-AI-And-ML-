# Feature Engneering
- df["column_name"].apply(func) : is used to apply a function on basis of specified column values and storing the output into new column
- df["column_name"].map(dictionary) : map the old value with a new value
- df.assign(new_income = df[column] * some changes or operations) : creates a new column(not written in quotes) of specified name by assigning the a specified value
- df["column_name"].replace(old, new) : replace specified old data of the column to new data of the column

```python
import pandas as pd
```

```python
df = pd.read_csv("4_4_persons_raw_data.csv")
df2 =  df.copy()
# lambda function are written with lambda then a variable with colon then body of function is defined note value for if conditon is written before if
df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 60000 else "10%")
```

```python
gender_map = {
    "Male" : "M", "Female" : "F", "Unknown":"U"
}
# gender_map
df2["gender"] =  df2["gender"].map(gender_map)    # maps the old value to new value
```

```python
df2 = df2.assign(new_income = df['income']*1.1)     # creates a new column with the specified name(not written string)by assigning the specified value 
```

```python
df2["country"] = df2["country"].replace("USA", "US")  # replace the old value with the new value
```
