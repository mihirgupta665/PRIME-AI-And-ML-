# Feature Engneering
- df["column_name"].apply(func) : is used to apply a function on basis of specified column values and storing the output into new column
- df["column_name"].map(dictionary) : map the old value with a new value
- df.assign(new_income = df[column] * some changes or operations) : creates a new column(not written in quotes) of specified name by assigning the a specified value
- df["column_name"].replace(old, new) : replace specified old data of the column to new data of the column

```python
import pandas as pd
```

```python
df = pd.read_csv("4_4_persons_raw_data.csv")
df2 =  df.copy()
# lambda function are written with lambda then a variable with colon then body of function is defined note value for if conditon is written before if
df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 60000 else "10%")
```

```python
gender_map = {
    "Male" : "M", "Female" : "F", "Unknown":"U"
}
# gender_map
df2["gender"] =  df2["gender"].map(gender_map)    # maps the old value to new value
```

```python
df2 = df2.assign(new_income = df['income']*1.1)     # creates a new column with the specified name(not written string)by assigning the specified value 
```

```python
df2["country"] = df2["country"].replace("USA", "US")  # replace the old value with the new value
```

## Data Transformation
- df2.columns = ["...columns_names..."] : changes all column values with the hard coded list
- df2.rename(columns={"column_old_name":"column_new_name"}) : rename the column with specified column and their name through mapping with a  dictionary
- df2.rename(index={old_vale:new_vale})  :  index could also be changed with mapping dictionary specifying old value and new value
- df2.[column_ name].replace("old_name" : "new_name") : replaces the row name to new row name for a specified column
- df2.sort_values(column_name, ascending=True) or df2.sort_values([...column_names...], ascending=False) :
  sort the entire table on the basic of specified column (if columns then first to last preferences)
- df2.reset_index(drop=False{optional}) : resets the indexing from 0 to last and also preserve the old indices by creating a new column named 'index', (drop=True) stop the preserving of old indices
- df2.sort_index() : sort the table according to the indices values
- df2["column"].rank() : gives ranking (lowest to highest) on the basis of specified column and in case of clash average is assigned
- df2["column"].rank(ascending=False, method="dense or min or max") : ascending=True gives rank from 1 to n And method=dense make ranking in whole numbers and no gap should be there if method is 'dense'.
- reorder columns :  df[[column_reorder_list]]  : changes the order of columns.

```python
df2.columns = ["ID", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"]   # hard coding the changed name in a list
df2 = df2.rename(columns={"Income":"Salary", "New_Income":"New_Salary"})   # changs the column names through mapping with a dictionary
df2.rename(index={1:"First"})   # df.rename() : could change column and index only
```

```python
df2['Name'] = df2['Name'].fillna("Mihir Gupta")
df2['Name'] = df2['Name'].replace("Jane Smith", "Joey Smith")
```

```python
df2
```

```python
df2['Salary'] = df2['Salary'].fillna(df2['Salary'].mean())
df2['New_Salary'] = df2['New_Salary'].fillna(df2['New_Salary'].mean())
df2.sort_values("Salary")
df2['Age'] = df2['Age'].fillna(df2['Age'].mean())
df2 = df2.sort_values(["Salary", "Age"], ascending=True)
```

```python
df2.reset_index()   # resets the index from 0 to n and preseves the old indices in a new index column
df2 = df2.reset_index(drop=False)   # if drop=True the new index column for preserving the index is droped
```

```python
df2
```

```python
df2['Gender']= df2['Gender'].fillna('F')
df2.sort_index()    # sort the index
```

```python
df2["ranking"] = df2["Salary"].rank()   # gives ranks accroding to the specified column and clashes are given average of logical ranks
df2["ranking"] = df2["Salary"].rank(ascending=True, method="dense")   # dense gives the whole number and maintains no gap in ranking
```
