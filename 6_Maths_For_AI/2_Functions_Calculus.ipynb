```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 100, 1000)
y = np.sqrt(x)

plt.plot(x, y, color="green", label="x-axis")
plt.title("sqrt of Input Values(X)")
plt.xlabel("X axis input  values")
plt.ylabel("Y  axis outplut values ")
plt.grid(True)
plt.legend()
plt.show()

```

```python
x = np.linspace( -4 * np.pi, 4 * np.pi, 100 )
y = np.sin(x)

plt.figure(figsize=(12,6))
plt.plot(x, y, color="yellow", label="sin(x)")
plt.axvline(0, color="black", linewidth=1.5, linestyle="--", label="X = 0")
plt.axhline(0, color="black", linewidth=1.5, linestyle="--", label="Y = 0")

plt.title("Sin(x) plot")
plt.xlabel("Input Values ")
plt.ylabel(" Output Sin Values ")
plt.legend()
plt.grid(True)
plt.show()
```
