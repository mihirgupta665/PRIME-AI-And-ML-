# Binomial Distribution

```python
from scipy.stats import binom

n=5  # total number of trials
k=3  # total number of success
p=0.5  # probability of one success

prob = binom.pmf(k,n,p)
print(prob)
```

# Uniform Distribution

- density = true : now y axis does not represent count rather it represents the likelyhood of a number falling in that particular bin

### linspace generates evenly spaced deterministic values in a range, whereas random.uniform generates random values drawn from a uniform probability distribution within that range.

```python
import numpy as np
import matplotlib.pyplot as plt

# generating values from 0 to 10
values = np.random.uniform(0, 10, 10_00_000)

# plotting a histgram becomes more  uniform with more values
plt.hist(values, bins=100, density=True, alpha=0.3)

plt.title("Representation of a uniform distibution between 0 to 10")
plt.xlabel("x")
plt.ylabel("Probability Distribution")
plt.grid(True)

plt.show()
```

# Normal Distribution
- It is a continous probability distribution that is bell-shaped and symmetric.
- Mean = Median = Mode
- 68% of data lies between (-1 sigma to +1 sigma) and 95% of data lies between (-2 sigma and +2 sigma) and 97% lies between (-3 sigma and +3 sigma)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(30, 110, 1000)

mean = 70
sigma = 10

# Normal Probalibility function
y = norm.pdf(x, mean, sigma)

plt.figure(figsize=(10,6))

plt.plot(x, y, color="black", linewidth=2, label="Normal Distribution")

x3 = np.linspace(mean - 3*sigma, mean + 3*sigma, 1000)
y3 = norm.pdf(x3, mean, sigma)
plt.fill_between(x3, y3, color="red", alpha=0.3, label="99.7% region (3 sigma)")

x2 = np.linspace( mean - 2*sigma, mean + 2*sigma, 1000 )
y2 = norm.pdf(x2, mean, sigma)
plt.fill_between(x2, y2, color="yellow", alpha=0.3, label="95% region (2 sigma)")

x1 = np.linspace( mean - sigma, mean + sigma, 1000)
y1 = norm.pdf(x1, mean, sigma)
plt.fill_between(x1, y1, color="green", alpha=0.3, label="68% region (1 sigma) ")

plt.title("Normal Distribution")
plt.xlabel("Scores (x -axis)")
plt.ylabel("Probability Density")
plt.legend()
plt.show()

```

- The mean of the sample mean will be equal to the population mean
- The standard deviation of the sample mean is the population standard deviation divided by the square root of sample size (sigma / sqrt(n))
