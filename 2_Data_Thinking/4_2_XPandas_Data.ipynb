# DataFrame Methods(Operations)
- df.head(n) : print n rows of data (by defualt 5 rows if n is missing)
- df.tail(n) : print n rows of data (by defualt 5 rows if n is missing)
- df.sample(n) : print each time n random row of the data (by default 1 row if n is missing)
- df.info() : gives the summary of : df data type, columns, columns datatype, memory usage
- df.shape : prints a tuple of configuration of shape of data i.e ( rows, columns )
- df.describe() : prints a numerical statistical summary  of all the numerical columns.
  stats visible are : count, mean, std, min, max,25%, 50%, 75%(25% means this is the data where 25% of values are less than or equal to) 
- df.columns : prints all the columns name
- df.nunique : prints the number of unique row in each columns
