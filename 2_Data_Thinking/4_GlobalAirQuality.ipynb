# Data Selection
- view is returned (not a copy) so changes in view will cause permanent changes in dataset too.
## Column
- df[column_name] : prints the entire specified column
- df[[column1, column2, ... , etc]] : prints the all the column entire data
## Row
- df.loc[label_index] : print the row of the specified label
- df.iloc[index] : print the row of the specified index
- Note if index and labbel are same (by default convention) same data will be printed.
- df.loc[0:n] : print the data from 0 index to n index(inclusive) 
- df.iloc[0:n] : print the data from 0 index to n index(exclusive)
## Cell
- df.at[row_label, col_label] : print the row and column label
- df.iat[row_index,col_index]
## Sub Data or Sub Table
- df.loc[row_label range, col_label range] : print the sub data(sub table) with the specified range
- df.iloc[row_index range, col_index range] : print the sub data(sub table) with the specified range

```python
df["city"]
```

```python
df[["city", "aqi"]]
```

```python
df.loc[2]
```

```python
df.iloc[2]
```

```python
df.loc[0:2]   # row 0,1,2(inclusice) will be printed
```

```python
df.iloc[0:2] # row 0,1   [2 exclusive ] will be printed
```

```python
df.at[0,"aqi"]
```

```python
df.iat[0,11]
```

```python
df.loc[0:2, ['city', 'aqi']]    # all values are inclusive  printing sub tables
```

```python
df.iloc[0:2, 1:5]  # sub tables using iloc always only accept the traditional numeric indices, last end range is exclusive 
```

# Data Filtering
- df[df['column'] condition (for eg. > 100)]   :  column is passed and a condition is imposed to get selected columns
- df[(condition) & | ~ (conditions) ... ]   :  multiple condition are passed with & | ~ operators
- 
- df[(condition) & | ~ (conditions)][column, column, column, ....]     :  filtering could chain with columns required to be extracted
- index or positioning resets after filtering but label is always preserved 
  

```python
df[df['aqi']>100]
```

```python
df[df["aqi"]>100]   # applying condition to column aqi 
```

```python
df[(df["aqi"]>100) & (df["temperature"]>30)]   # combining condition with & | ~ operator 
```

```python
# extracting specific desired column by chaing columns after the filtering conditions
aqi100_temp30 = df[(df["aqi"]>100) & (df["temperature"]>30)][["city","aqi"]]
aqi100_temp30
```

```python
aqi100_temp30.iloc[0]   # index changes and positioning resets start from 0 again
```

```python
# index changes and positioning resets start from 0 again
# label is preserved after filtering(extraction) 
# aqi100_temp30.loc[0] will give error  as it was not filtered 
aqi100_temp30.loc[6]   # first row extracted had the label 6 and would be preserved too always .
```

## Data Selection Query
- df.query(Expression in string ][filtering columns if required)     [chain comparision in expression]
- Expression are string and column could directly be referenced in the expression
- use backtick if column_name has space or special characters
- Operators like (& | < >= etc) could directly be written in the string expression
- @var : variable could be passed with a prefix of '@'
- Note : the data returned is a copy of data (not a view as in loc or iloc or at or iat)

```python
aqi_val = 100
# df.query(expression) : is  used to extract data on basic of conditional filtering
df.query("aqi > @aqi_val")    # variable is written with '@' as prefix
```

```python
df.query("aqi > 100 & temperature >30")[['city', 'aqi']]  # chaining and opertors could be used for enhanced filtering
```
